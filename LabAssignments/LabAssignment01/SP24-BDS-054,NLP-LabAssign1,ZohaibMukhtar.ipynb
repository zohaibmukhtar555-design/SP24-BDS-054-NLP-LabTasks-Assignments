## Task 1: Regex Fundamentals (Character Classes & Quantifiers)

Given a sample text with mixed content (words, numbers, symbols), write regex patterns for:

a) Extract all digits from the text
b) Extract all words that start with a capital letter
c) Extract all sequences of exactly 4 digits (e.g., years)
d) Extract all words that are between 3 to 6 characters long


```python
# sample text with mixed words, numbers and symbols

sample_text = "In 2024, Ahmed Khan bought 3 items worth $250 and 15 books from Karachi. \
The order number was #4587 and delivery takes 2 days. Contact info: 03001234567, \
price 12.5% off. Sara2000 also visited in July 1999."

print(sample_text)
```

```python
# a) extract all digits

digits = re.findall(r'\d+', sample_text)
print("Digits found:", digits)
```

```python
# b) extract words that start with a capital letter

cap_words = re.findall(r'\b[A-Z][a-z]*\b', sample_text)
print("Capitalized words:", cap_words)
```

```python
# c) extract sequences of exactly 4 digits (like years)

four_digit = re.findall(r'\b\d{4}\b', sample_text)
print("4-digit sequences:", four_digit)
```

```python
# d) extract words that are 3 to 6 characters long

words_3_6 = re.findall(r'\b[a-zA-Z]{3,6}\b', sample_text)
print("3-6 char words:", words_3_6)
```

## Task 2: Pattern-Based Extraction (Real World Data)

Given a paragraph containing emails, phone numbers, order IDs, dates, and percentages mixed
together:

a) Write a regex pattern using `re.findall()` to extract all email addresses
b) Write a regex pattern to extract all phone numbers (assume format: 03XX-XXXXXXX)
c) Write a regex pattern to extract all dates in DD/MM/YYYY format
d) Write a regex pattern to extract all percentage values (e.g., 45%, 12.5%)


```python
# paragraph with emails, phone numbers, order ids, dates and percentages mixed in

paragraph2 = """Please contact us at support@shopnow.com or sales.team@business.org for any
queries. You can also reach the manager at ali.raza99@gmail.com. Our customer helpline numbers
are 0301-2345678 and 0333-9998887. Your order #ORD1234 was placed on 15/08/2024 and is expected
to arrive by 22/08/2024. Note that a discount of 20% applies on this order, while returning
customers get an extra 12.5% off. Another order, #ORD5678, was placed on 01/01/2023 with a
flat 50% discount."""

print(paragraph2)
```

```python
# a) extract all email addresses

emails = re.findall(r'[a-zA-Z0-9._]+@[a-zA-Z0-9.-]+\.[a-zA-Z]+', paragraph2)
print("Emails found:", emails)
```

```python
# b) extract phone numbers in 03XX-XXXXXXX format

phones = re.findall(r'03\d{2}-\d{7}', paragraph2)
print("Phone numbers found:", phones)
```

```python
# c) extract dates in DD/MM/YYYY format

dates = re.findall(r'\d{2}/\d{2}/\d{4}', paragraph2)
print("Dates found:", dates)
```

```python
# d) extract percentage values

percentages = re.findall(r'\d+(?:\.\d+)?%', paragraph2)
print("Percentages found:", percentages)
```

## Task 3: Word Tokenization (NLTK vs spaCy)

Using a given paragraph of your choice (at least 3-4 sentences, including contractions like
"don't", "it's"):

a) Tokenize the paragraph using NLTK's `word_tokenize()`
b) Tokenize the same paragraph using a blank spaCy pipeline
c) Compare and write down (in a comment) at least 2 differences you noticed between the two
outputs
d) Count the total number of tokens produced by each method


```python
# my own paragraph with contractions and a hyphenated word

paragraph3 = "I don't think it's a good idea to skip breakfast, especially before a long day. \
My well-known friend's car broke down near 5th Avenue yesterday, costing about $1,200 to fix. \
We'll try again this weekend, though!"

print(paragraph3)
```

```python
# a) tokenize using NLTK

nltk_tokens = word_tokenize(paragraph3)
print("NLTK tokens:")
print(nltk_tokens)
```

```python
# b) tokenize using a blank spaCy pipeline (no trained model, just the tokenizer)

nlp_blank = spacy.blank("en")
doc3 = nlp_blank(paragraph3)
spacy_tokens = [token.text for token in doc3]

print("spaCy tokens:")
print(spacy_tokens)
```

```python
# c) comparing the two outputs
# Difference 1: NLTK keeps "well-known" as ONE single token, but spaCy splits it into
#               three separate tokens -> "well", "-", "known"
# Difference 2: NLTK writes the number as "1,200" (comma stays attached), spaCy also keeps
#               "1,200" together here, but in general spaCy handles punctuation attached to
#               words a bit differently which is why the hyphenated word split happened
# both of them handled "don't" -> "do" + "n't" and "it's" -> "it" + "'s" the exact same way
```

```python
# d) total token count from each method

print("Total tokens from NLTK:", len(nltk_tokens))
print("Total tokens from spaCy:", len(spacy_tokens))
```

## Task 4: Token Attributes (Categorizing Tokens)

Using a sample sentence that contains normal words, numbers, an email address, and a currency
symbol (e.g., "$50" or "Rs. 500"):

a) Load the sentence into a spaCy Doc object
b) Loop through each token and print its `is_alpha` value
c) Print each token's `like_num` and `like_email` values
d) Identify and list only the tokens where `is_currency` is True


```python
# a) load sentence into a spaCy Doc (using the trained pipeline this time)

nlp = spacy.load("en_core_web_sm")

sentence4 = "I paid $50 for 2 books and contacted support at help@bookstore.com for Rs. 500 refund."
doc4 = nlp(sentence4)

print(doc4)
```

```python
# b) print is_alpha for each token

for token in doc4:
    print(token.text, "-> is_alpha:", token.is_alpha)
```

```python
# c) print like_num and like_email for each token

for token in doc4:
    print(token.text, "-> like_num:", token.like_num, ", like_email:", token.like_email)
```

```python
# d) tokens where is_currency is True

currency_tokens = []
for token in doc4:
    if token.is_currency:
        currency_tokens.append(token.text)

print("Currency tokens:", currency_tokens)
# note: "Rs" doesn't get flagged as currency here because spaCy splits "Rs." into "Rs" + "."
# and "Rs" on its own isn't recognized as a currency symbol, only "$" is
```
