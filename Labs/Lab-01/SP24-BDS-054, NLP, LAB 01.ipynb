## Task 04: Sets for Unique Vocabulary

```python
# (a) Convert list with duplicates to set[cite: 11]
raw = ["nlp", "ai", "nlp", "deep", "learning", "ai", "nlp", "python"]
vocab = set(raw)
print("(a) Raw list:", raw)
print("    Vocab set:", vocab)
print("    Number of unique words:", len(vocab))

# Explanation comment
# Sets automatically remove duplicates because Python sets are hash tables.
# Each distinct element corresponds to a unique hash key, preventing multiple identical keys
```

```python
# (b) Set operations: union, intersection, difference[cite: 11]
set_A = {"machine", "learning", "data", "model"}
set_B = {"deep", "learning", "neural", "data"}

print("\n(b) Union:", set_A | set_B)
print("    Intersection:", set_A & set_B)
print("    Difference (set_A - set_B):", set_A - set_B)
```

```python
# (c) Removing stopwords via set difference[cite: 11]
stopwords = {"is", "the", "a", "in", "of", "and", "to"}
tokens = ["nlp", "is", "a", "field", "of", "ai", "and", "the", "future"]

# Preserving order while filtering with set difference[cite: 11]
cleaned_tokens = [t for t in tokens if t in (set(tokens) - stopwords)]
print("\n(c) Cleaned tokens (stopwords removed):", cleaned_tokens)
```

```python
# (d) Add, discard, and membership check[cite: 11]
vocab.add("transformer")
vocab.add("bert")
vocab.discard("python")

print("\n(d) Is 'ai' present in vocab?:", "ai" in vocab)
print("    Final vocab set:", vocab)
```

## Task 06: String Cleaning Methods

```python
# (a) Sequential cleaning[cite: 11]
text = "   MACHINE Learning is AMAZING!!!   "
step1 = text.strip()
print("(a) After strip():", repr(step1))
step2 = step1.lower()
print("    After lower():", step2)
step3 = step2.replace("!", "")
print("    After replace('!', ''):", step3)
```

```python
# (d) Normalization function and test runs[cite: 11]
def normalize(text_input):
    cleaned = text_input.strip().lower()
    cleaned = cleaned.replace(",", "").replace(".", "")
    return cleaned

test_1 = "   Natural Language, Processing with Python.   "
test_2 = "   Hello, world. Machine Learning is awesome.   "

print("\n(d) Test 1 normalized:", normalize(test_1))
print("    Test 2 normalized:", normalize(test_2))
```

## Task 07: String Validation Methods

```python
# (a) Prefix, suffix, and substring location[cite: 11]
text = "NLP is the future of AI research"
print("(a) Starts with 'NLP':", text.startswith("NLP"))
print("    Ends with 'research':", text.endswith("research"))
print("    Starting index of 'future':", text.find("future"))
```

```python
# (b) Loop classification of mixed list[cite: 11]
tokens = ["hello", "123", "world", "456abc", "python", "789", "NLP3"]
print("\n(b) Token classifications:")
for t in tokens:
    if t.isalpha():
        label = "Purely Alphabetic"
    elif t.isdigit():
        label = "Purely Numeric"
    else:
        label = "Mixed"
    print(f"    '{t}' -> {label}")
```

```python
# (c) Segregating list into three target buckets[cite: 11]
words = ["running", "jumped", "NLP", "ai2025", "data", "42"]
pure_words = []
pure_numbers = []
mixed_tokens = []

for w in words:
    if w.isalpha():
        pure_words.append(w)
    elif w.isdigit():
        pure_numbers.append(w)
    else:
        mixed_tokens.append(w)

print("\n(c) pure_words:", pure_words)
print("    pure_numbers:", pure_numbers)
print("    mixed_tokens:", mixed_tokens)
```

```python
# (d) Modular token validation function[cite: 11]
def validate_token(token):
    if token.isalpha():
        return "WORD"
    elif token.isdigit():
        return "NUMBER"
    return "MIXED"

test_candidates = ["chat", "99", "gpt4", "bert", "2024", "nlp!"]
print("\n(d) Validated tokens:")
for t in test_candidates:
    print(f"    {t} -> {validate_token(t)}")
```

## Task 08: Functions and Lambda for NLP Pipelines

```python
# (a) Regular function vs Lambda equivalent[cite: 11]
def clean(text):
    return text.strip().lower()

clean_lambda = lambda text: text.strip().lower()

sample_input = "  HELLO Natural Language Processing  "
print("(a) Regular function:", repr(clean(sample_input)))
print("    Lambda function: ", repr(clean_lambda(sample_input)))
```

```python
# (b) Tokenizer and filtering short tokens[cite: 11]
def tokenize(text):
    return text.split()

def remove_short(tokens, n):
    return [t for t in tokens if len(t) > n]

target_sentence = "I love working with NLP tools in Python daily"
chained_tokens = remove_short(tokenize(target_sentence), 3)
print("\n(b) Chained output (length > 3):", chained_tokens)
```

```python
# (c) Dictionary of token lengths[cite: 11]
def word_lengths(tokens):
    return {w: len(w) for w in tokens}

sample_tokens = ["language", "model", "ai", "processing", "deep"]
print("\n(c) Word lengths dict:", word_lengths(sample_tokens))
```

```python
# (d) Full end-to-end NLP pipeline[cite: 11]
def nlp_pipeline(text, min_len):
    cleaned = clean(text)
    token_list = tokenize(cleaned)
    filtered = remove_short(token_list, min_len)
    return word_lengths(filtered)

sentence_1 = "BERT and GPT models are revolutionizing modern AI architectures"
sentence_2 = "Text preprocessing is an essential first step in machine learning pipelines"

print("\n(d) Pipeline output 1:\n   ", nlp_pipeline(sentence_1, 3))
print("    Pipeline output 2:\n   ", nlp_pipeline(sentence_2, 4))
```

## Task 09: map(), filter(), sorted(), and Counter

```python
from collections import Counter

words = ["Python", "NLP", "is", "a", "powerful", "tool", "for", "text"]

# (a) map() applications[cite: 11]
lowercased = list(map(lambda w: w.lower(), words))
word_lengths = list(map(lambda w: len(w), words))
print("(a) Lowercased with map():", lowercased)
print("    Lengths with map():", word_lengths)
```

```python
# (b) filter() applications[cite: 11]
words_long = list(filter(lambda w: len(w) > 3, words))
words_with_t = list(filter(lambda w: "t" in w.lower(), words))
print("\n(b) Words > 3 chars:", words_long)
print("    Words containing 't':", words_with_t)
```

```python
# (c) sorted() variants[cite: 11]
print("\n(c) (i) Alphabetical:", sorted(words))
print("    (ii) By length ascending:", sorted(words, key=len))
print("    (iii) By length descending:", sorted(words, key=len, reverse=True))
```

```python
# (d) Frequency counting using collections.Counter[cite: 11]
text_corpus = "nlp nlp python ai nlp python text ai nlp python text"
tokens = text_corpus.split()
freq = Counter(tokens)

print("\n(d) Full Counter:", freq)
print("    (i) 3 most common words:", freq.most_common(3))
print("    (ii) Frequency of 'python':", freq["python"])
print("    (iii) Total words:", len(tokens))
print("          Unique vocabulary size:", len(freq))
```
